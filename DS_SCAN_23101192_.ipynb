```python
#Left
head = int(input("Enter initial head position: "))

request = list(map(int, input("Enter requests: ").split()))

left = sorted(filter(lambda x: x < head, request))
right = sorted(filter(lambda x: x > head, request))

total = 0

for i in left:
    total += abs(head - i)
    head = i

for i in right:
    total += abs(head - i)
    head = i

print(total)
```

```python
# Elevator Scheduling - Right

head = int(input("Enter head: "))
requests = list(map(int, input("Enter requests: ").split()))

right = sorted([x for x in requests if x > head])
left = sorted([x for x in requests if x < head], reverse=True)

total = 0

for x in right:
    total += abs(head - x)
    head = x

for x in left:
    total += abs(head - x)
    head = x

print("Total Head Movement =", total)
```
