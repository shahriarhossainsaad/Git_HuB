```python
class Process:
    def __init__(self, pid, at, bt):
        self.pid = pid
        self.at = at
        self.bt = bt
        self.rt = bt
        self.ct = 0
        self.tat = 0
        self.wt = 0


n = int(input("Enter the number of processes: "))
p = []

for i in range(n):
    at = int(input(f"P{i+1} Arrival Time: "))
    bt = int(input(f"P{i+1} Burst Time: "))
    p.append(Process(f"P{i+1}", at, bt))


time = 0
done = 0
tq = 2
queue = []

while done < n:

    # Add arrived processes to queue
    for x in p:
        if x.at <= time and x.rt > 0 and x not in queue:
            queue.append(x)

    if not queue:
        time += 1
        continue

    cur = queue.pop(0)

    # Run for maximum 2 units
    run_time = min(tq, cur.rt)

    cur.rt -= run_time
    time += run_time

    # Add newly arrived processes
    for x in p:
        if x.at <= time and x.rt > 0 and x not in queue and x != cur:
            queue.append(x)

    if cur.rt == 0:
        cur.ct = time
        cur.tat = cur.ct - cur.at
        cur.wt = cur.tat - cur.bt
        done += 1
    else:
        # Put current process at the end
        queue.append(cur)


print("\nPID\tAT\tBT\tCT\tTAT\tWT")

for x in p:
    print(x.pid, x.at, x.bt, x.ct, x.tat, x.wt, sep="\t")

print("\nAverage TAT:", sum(x.tat for x in p) / n)
print("Average WT:", sum(x.wt for x in p) / n)
```
