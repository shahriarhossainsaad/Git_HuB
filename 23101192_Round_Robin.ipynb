```python
from collections import deque

class Process:
    def __init__(self, pid, at, bt):
        self.pid = pid
        self.at = at
        self.bt = bt
        self.rt = bt
        self.ct = 0
        self.tat = 0
        self.wt = 0


n = int(input("Enter the number of processes: "))
p = []

for i in range(n):
    at = int(input(f"P{i+1} Arrival Time: "))
    bt = int(input(f"P{i+1} Burst Time: "))
    p.append(Process(f"P{i+1}", at, bt))

p.sort(key = lambda x:x.at)

q = deque()
time = 0
done = 0
tq = 5


while done < n:


    for process_in_p in p:
        if process_in_p.at <= time and process_in_p.rt > 0 and process_in_p not in q:
            q.append(process_in_p)

    if not q:
        time += 1
        continue

    cur = q.popleft()


    run = min(tq, cur.rt)

    cur.rt -= run
    time += run


    for process_in_p in p:
        if process_in_p.at <= time and process_in_p.rt > 0 and process_in_p not in q and process_in_p != cur:
            q.append(process_in_p)

    if cur.rt == 0:
        cur.ct = time
        cur.tat = cur.ct - cur.at
        cur.wt = cur.tat - cur.bt
        done += 1
    else:

        q.append(cur)


print("\nPID\tAT\tBT\tCT\tTAT\tWT")

for x in p:
    print(x.pid, x.at, x.bt, x.ct, x.tat, x.wt, sep="\t")

print("\nAverage TAT:", sum(x.tat for x in p) / n)
print("Average WT:", sum(x.wt for x in p) / n)
```
