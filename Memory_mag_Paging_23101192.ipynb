```python

from collections import deque

pages = list(map(int, input("Enter reference string: ").split()))
frame_size = int(input("Enter frame size: "))

queue = deque()
page_fault = 0
page_hit = 0

for page in pages:

    if page in queue:
        page_hit += 1

    else:
        page_fault += 1

        if len(queue) == frame_size:
            queue.popleft()

        queue.append(page)

    print(page, "->", list(queue))

hit_ratio = page_hit / len(pages)

print("\nPage Fault =", page_fault)
print("Page Hit =", page_hit)
print("Page Hit Ratio =", hit_ratio)
```
